### Further Exploratory Data Analysis (EDA)

Let's visualize the distribution of the target variable (`MedHouseVal`) and explore its relationship with some key features, including a geographical view.

````python
# Distribution of the target variable: MedHouseVal
plt.figure(figsize=(10, 6))
sns.histplot(y['MedHouseVal'], bins=50, kde=True)
plt.title('Distribution of Median House Value (MedHouseVal)')
plt.xlabel('Median House Value ($100,000s)')
plt.ylabel('Frequency')
plt.show()
````

The distribution of `MedHouseVal` is skewed to the left, with a peak around $150,000-$200,000. There's also a noticeable cap at $500,000, which is a common characteristic of this dataset, indicating that values above this were capped during data collection.

````python
# Relationship between selected features and the target variable

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 12))
axes = axes.flatten()

features_to_plot = ['MedInc', 'HouseAge', 'AveRooms', 'Population']

for i, feature in enumerate(features_to_plot):
    sns.scatterplot(x=X[feature], y=y['MedHouseVal'], ax=axes[i], alpha=0.6)
    axes[i].set_title(f'{feature} vs. MedHouseVal')
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel('Median House Value ($100,000s)')

plt.tight_layout()
plt.show()
````

From these scatter plots, we can observe:
- **MedInc (Median Income)**: There's a strong positive correlation between `MedInc` and `MedHouseVal`. Higher median incomes generally correspond to higher house values.
- **HouseAge**: The relationship with `HouseAge` appears less direct, though older houses might be concentrated in certain value ranges.
- **AveRooms (Average Rooms)**: A positive correlation is visible, indicating that properties with more average rooms tend to have higher values.
- **Population**: `Population` doesn't seem to have a clear linear relationship with `MedHouseVal` on its own.

### Geospatial Visualization of Median House Value

Let's visualize the median house value across California using Latitude and Longitude. This can reveal regional patterns in housing prices.

````python
# Combine X and y for easier plotting
df_combined = X.copy()
df_combined['MedHouseVal'] = y['MedHouseVal']

plt.figure(figsize=(10, 10))
sns.scatterplot(
    data=df_combined,
    x='Longitude',
    y='Latitude',
    hue='MedHouseVal',
    size='MedHouseVal',
    palette='viridis',
    alpha=0.4,
    sizes=(20, 400) # Adjust size range for better visualization
)
plt.title('Median House Value across California (Geospatial)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
````

The geospatial plot clearly shows clusters of higher median house values (represented by brighter colors and larger points) in certain regions of California, particularly along the coast and in major metropolitan areas, as expected. This suggests that location (Latitude and Longitude) is a very important factor in determining housing prices.
